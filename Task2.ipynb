```python
import xarray as xr
```

```python
# Use the GCS path, not the Cloud Console browser URL (that URL returns HTML, not Zarr metadata).
zarr_path = "gs://gcp-public-data-arco-era5/ar/1959-2022-6h-1440x721.zarr"
```

```python
# Open the dataset using Xarray.
ds = xr.open_zarr(zarr_path, storage_options={"token": "anon"})
```

```python
# 120-hour window of 10 m wind speed (21 timesteps at 6-hourly resolution)
import numpy as np

end_time = "2021-12-31T18:00:00"
start_time = np.datetime64(end_time) - np.timedelta64(120, "h")

measurement = "10m_u_component_of_wind"
wind = ds[measurement].sel(time=slice(start_time, end_time))
# Pull the slice into memory once (avoids re-fetching from GCS on every animation frame)
wind = wind.load()

print(wind.dims, wind.shape)
print(wind.time.values[0], "→", wind.time.values[-1])
```

```python
# Animated map of wind speed over the 120-hour period

import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Optional: subsample grid for faster drawing (full grid is 1440 x 721)
stride = 2
lon = wind.longitude.values[::stride]
lat = wind.latitude.values[::stride]
frames = wind.values[:, ::stride, ::stride]

vmin, vmax = np.nanpercentile(frames, [2, 98])

fig, ax = plt.subplots(figsize=(12, 4.5))
# imshow: rows = latitude (90°N → 90°S), cols = longitude
im = ax.imshow(
    frames[0],
    origin="upper",
    extent=[lon.min(), lon.max(), lat.min(), lat.max()],
    vmin=vmin,
    vmax=vmax,
    cmap="viridis",
    aspect="auto",
)
cb = fig.colorbar(im, ax=ax, pad=0.02)
cb.set_label("10 m u-component of wind (m s⁻¹)")
ax.set_xlabel("Longitude (°E)")
ax.set_ylabel("Latitude (°N)")
title = ax.set_title("")

def update(i):
    im.set_data(frames[i])
    t = np.datetime_as_string(wind.time.values[i], unit="h")
    title.set_text(f"10 m u-component of wind — {t} UTC")
    return im, title

anim = FuncAnimation(fig, update, frames=len(wind.time), interval=500, blit=False)
plt.close(fig)

# Inline playback in the notebook
HTML(anim.to_jshtml())
```

Running the above animation, we notice the wind appears to blow horizontally. This makes sense, given the choice of our variable. We can also see different wind speeds travelling across the Globe, for instance, around -60°.
